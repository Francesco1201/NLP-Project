# Analysis of the scores

Builds summary tables with all the results: one by model, one by game category (all models together) and one by game category for each model, from the scores in `scores/` (exported from the evaluation workbooks). The tables are saved in `tables/` as CSV and as LaTeX tables (`full_results`, `category_results` and `category_results_<model>`).

- Criteria on a 1-5 scale: mean per model.
- Criteria on a 0-2 scale: percentage of answers with score 2 / 1 / 0.
- Flaw Detection: recall (share of injected flaws detected), also by type of flaw; Type correct and fixes computed on detected flaws; invented problems as mean per answer.

Models are listed by size.

```python
import os, sys
if "google.colab" in sys.modules:
    from google.colab import drive
    drive.mount("/content/drive")
    os.chdir("/content/drive/My Drive/Progetto NLP 2026/evaluation")
sys.path.insert(0, os.getcwd())

import metrics

data = metrics.ScoreData("scores")
report = metrics.ReportWriter("tables")
counts = data.counts()
expected = {"explain": 45, "flawOutputs": 45, "flawItems": 135, "simulation": 45, "generation": 15, "variant": 15}
assert all((counts[name] == number).all() for name, number in expected.items()), "unexpected number of scores"
counts
```

## Results by model

```python
caption = ("Results of all criteria per task and model. Criteria on a 1--5 scale are means; criteria on a 0--2 scale "
           "are the percentage of answers with score 2/1/0. Recall is the share of injected flaws detected; "
           "Type correct and fixes are computed on detected flaws; invented problems are a mean per answer.")
fullResults = report.table(metrics.fullTable(data), "full_results")
report.latex(metrics.fullLatex(fullResults, caption, "tab:full_results"), "full_results")
fullResults
```

## Results by game category

The same criteria with the five game categories as columns, computed on all models together. Each cell is computed on 36 answers (Explain, Simulation), 108 injected flaws (Flaw Detection) or 12 answers (Generation of a New Game, Variant: one specification or base game per category), so the differences between categories are indicative only.

```python
caption = ("Results of all criteria per task and game category, all models together. Criteria on a 1--5 scale are means; "
           "criteria on a 0--2 scale are the percentage of answers with score 2/1/0. Recall is the share of injected flaws "
           "detected; Type correct and fixes are computed on detected flaws; invented problems are a mean per answer.")
categoryResults = report.table(metrics.fullTable(data, "category"), "category_results")
report.latex(metrics.fullLatex(categoryResults, caption, "tab:category_results"), "category_results")
categoryResults
```

## Results by game category, for each model

The same table by game category, computed separately for each model. Each cell is computed on 9 answers (Explain, Simulation), 27 injected flaws (Flaw Detection) or 3 answers (Generation of a New Game, Variant), so these tables are indicative only.

```python
for model in metrics.modelOrder:
    caption = (f"Results of all criteria per task and game category for {metrics.shortLabels[model]}. Criteria on a 1--5 scale "
               "are means; criteria on a 0--2 scale are the percentage of answers with score 2/1/0. Recall is the share of "
               "injected flaws detected; Type correct and fixes are computed on detected flaws; invented problems are a "
               "mean per answer.")
    fileName = "category_results_" + metrics.shortLabels[model].lower().replace(" ", "_").replace(".", "")
    modelResults = report.table(metrics.fullTable(data.forModel(model), "category"), fileName)
    report.latex(metrics.fullLatex(modelResults, caption, f"tab:{fileName}"), fileName)
    print(metrics.shortLabels[model])
    display(modelResults)
```
